```python
# exercicio 7
numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

positivos = []
negativos = []
zeros = []

for num in numeros:
    if num > 0:
        positivos.append(num)
    elif num < 0:
        negativos.append(num)
    else:
        zeros.append(num)

# Exibindo os resultados
print("Números positivos:", positivos)
print("Números negativos:", negativos)
print("Zeros:", zeros)
```

```python
# exercicio 8
import copy

# Lista inicial com 10 números escolhidos
lista_original = [42, 7, 15, -3, 88, 0, 23, 11, -12, 5]

# Desafio: fazendo uma cópia independente da lista original
lista_trabalho = lista_original.copy()

print(f"Lista Original:     {lista_original}")

lista_crescente = sorted(lista_trabalho)
print(f"Ordem Crescente:    {lista_crescente}")

lista_decrescente = sorted(lista_trabalho, reverse=True)
print(f"Ordem Decrescente:  {lista_decrescente}")

maior_valor = max(lista_original)
menor_valor = min(lista_original)

print(f"Maior valor:        {maior_valor}")
print(f"Menor valor:        {menor_valor}")

print(f"\nVerificação da lista original (intacta): {lista_original}")
```

```python
# exercicio 9
produtos = {
    "Arroz": 25.90,
    "Feijão": 8.50,
    "Macarrão": 6.90,
    "Leite": 5.50,
    "Café": 15.00
}
print("--- LISTA DE PRODUTOS ---")
for nome, preco in produtos.items():
    print(f"{nome:10} -> R$ {preco:.2f}")

total = sum(produtos.values())

mais_caro = max(produtos, key=produtos.get)
mais_barato = min(produtos, key=produtos.get)

print("\n--- RESUMO DA COMPRA ---")
print(f"Valor total:       R$ {total:.2f}")
print(f"Produto mais caro:   {mais_caro} (R$ {produtos[mais_caro]:.2f})")
print(f"Produto mais barato: {mais_barato} (R$ {produtos[mais_barato]:.2f})")

# Desafio: Busca direta no dicionário
print("\n--- DESAFIO: BUSCAR PRODUTO ---")
busca = input("Digite o nome de um produto para buscar: ").strip().capitalize()

if busca in produtos:
    print(f" Produto cadastrado! {busca} custa R$ {produtos[busca]:.2f}.")
else:
    print(f" O produto '{busca}' NÃO está cadastrado.")
```

```python
# exercicio 10
alunos = []

qtd = int(input("Quantos alunos serão cadastrados? "))

for i in range(1, qtd + 1):
    print(f"\n--- Cadastro do Aluno {i} ---")
    nome = input("Nome: ").strip().capitalize()
    nota = float(input("Nota (0 a 10): "))

    alunos.append({"nome": nome, "nota": nota})


soma_notas = sum(aluno["nota"] for aluno in alunos)
media_turma = soma_notas / qtd if qtd > 0 else 0

aluno_maior_nota = max(alunos, key=lambda aluno: aluno["nota"])
aluno_menor_nota = min(alunos, key=lambda aluno: aluno["nota"])

aprovados = [aluno for aluno in alunos if aluno["nota"] >= 7.0]
reprovados = [aluno for aluno in alunos if aluno["nota"] < 7.0]


print("\n" + "="*40)
print("          RELATÓRIO DA TURMA          ")
print("="*40)

print("\n--- Todos os Alunos ---")
for aluno in alunos:
    print(f"• {aluno['nome']:<15} -> Nota: {aluno['nota']:.1f}")

print("\n--- Estatísticas ---")
print(f"Média da turma:      {media_turma:.2f}")
print(f"Maior nota:          {aluno_maior_nota['nome']} ({aluno_maior_nota['nota']:.1f})")
print(f"Menor nota:          {aluno_menor_nota['nome']} ({aluno_menor_nota['nota']:.1f})")

print("\n--- Situação dos Alunos ---")
print("Aprovados (Nota >= 7):")
if aprovados:
    for a in aprovados:
        print(f"  [APROVADO]  {a['nome']} ({a['nota']:.1f})")
else:
    print("  Nenhum aluno aprovado.")

print("\nReprovados (Nota < 7):")
if reprovados:
    for r in reprovados:
        print(f"  [REPROVADO] {r['nome']} ({r['nota']:.1f})")
else:
    print("  Nenhum aluno reprovado.")


print("\n" + "="*40)
print("            PESQUISA DE ALUNO           ")
print("="*40)

busca = input("\nDigite o nome do aluno que deseja pesquisar: ").strip().capitalize()

encontrado = next((aluno for aluno in alunos if aluno["nome"] == busca), None)

if encontrado:
    situacao = "Aprovado" if encontrado["nota"] >= 7.0 else "Reprovado"
    print(f"\n Aluno encontrado!")
    print(f"   Nome:     {encontrado['nome']}")
    print(f"   Nota:     {encontrado['nota']:.1f}")
    print(f"   Situação: {situacao}")
else:
    print(f"\n O aluno '{busca}' não foi encontrado no cadastro.")
```
