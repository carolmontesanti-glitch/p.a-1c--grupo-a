```python
# exercicio 9

# crie uma lista vazia chamada numeros
# peça ao usuario p0ara informar 10 numeros
# depois crie 2 listas vazias chamadas pares e impares
# percorra a lista orginal e coloque cada numero na lista correspondente

num = []
pares = []
impar = []

for i in range(10):
  n = int(input(f"informe o {i+1} numero: "))
  num.append(n)
for n in num:
  if n % 2 == 0:
    pares.append(n)
  else:
    impar.append(n)
  print("")
  print(f"pares: {pares}")
  print(f"impares: {impar}")
```

```python
# exercicio 10

# faça um pequeno sistema para cadastrar as notas de 05 alunos
# o programa deverá armazenar as notas em uma lista e depois mostrar:
# todas as notas, a menor, a média, quantidade de alunos com nota > ou = a 6, quant de alunos com nota < 6.

notas = []
for i in range(1, 6):
  nota = float(input(f"Digite a nota do aluno {i}: "))
  notas.append(nota)

media = sum(notas) / len(notas)
aprovados = len([n for n in notas if n >= 6])
reprovados = len([n for n in notas if n < 6])

print("Resultado")
print(f"Todas as notas: {notas}")
print(f"Menor nota: {notas}")
print(f"Média da turma: {media}")
print(f"Alunos com nota maior ou igual a 6: {aprovados}")
print(f"Alunos com nota menor que 6: {reprovados}")
```
